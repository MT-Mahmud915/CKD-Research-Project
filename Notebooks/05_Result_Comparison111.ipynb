```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import os
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")

# ----------------------------------------------------------
# Plot Style (Publication Quality)
# ----------------------------------------------------------

plt.style.use("seaborn-v0_8-whitegrid")

plt.rcParams["figure.dpi"] = 300
plt.rcParams["savefig.dpi"] = 300

plt.rcParams["font.size"] = 11
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["axes.titlesize"] = 13

plt.rcParams["legend.fontsize"] = 10

# ----------------------------------------------------------
# Project Directories
# ----------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/CKD Research Project"

FIGURE_DIR = os.path.join(BASE_DIR, "Figures")
RESULT_DIR = os.path.join(BASE_DIR, "Results")
MODEL_DIR = os.path.join(BASE_DIR, "Saved_Models")

os.makedirs(FIGURE_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)

# ----------------------------------------------------------
# Result Files
# ----------------------------------------------------------

ML_RESULT_FILE = os.path.join(
    RESULT_DIR,
    "Machine_Learning_Model_Comparison.csv"
)

DNN_RESULT_FILE = os.path.join(
    RESULT_DIR,
    "DNN_Results.csv"
)

print("="*60)
print("CKD Research Project")
print("Notebook 05 : ML vs DNN Comparison")
print("="*60)

print(f"\nFigure Directory : {FIGURE_DIR}")
print(f"Result Directory : {RESULT_DIR}")
print(f"Model Directory  : {MODEL_DIR}")

print("\nNotebook Initialized Successfully.")
```

```python
# ==========================================================
# Section 2 : Load Machine Learning & DNN Results
# ==========================================================

import os
import pandas as pd

# ----------------------------------------------------------
# Result Files
# ----------------------------------------------------------

ML_RESULT_FILE = os.path.join(
    RESULT_DIR,
    "model_comparison.csv"
)

DNN_RESULT_FILE = os.path.join(
    RESULT_DIR,
    "DNN_Performance.csv"
)

# ----------------------------------------------------------
# Check Files
# ----------------------------------------------------------

if not os.path.exists(ML_RESULT_FILE):
    raise FileNotFoundError(f"ML result file not found:\n{ML_RESULT_FILE}")

if not os.path.exists(DNN_RESULT_FILE):
    raise FileNotFoundError(f"DNN result file not found:\n{DNN_RESULT_FILE}")

# ----------------------------------------------------------
# Load Results
# ----------------------------------------------------------

ml_results = pd.read_csv(ML_RESULT_FILE)
dnn_results = pd.read_csv(DNN_RESULT_FILE)

# ----------------------------------------------------------
# Display Basic Information
# ----------------------------------------------------------

print("=" * 70)
print("Machine Learning Results")
print("=" * 70)

print("Shape :", ml_results.shape)
print("\nColumns:")
print(list(ml_results.columns))

display(ml_results)

print("\n")

print("=" * 70)
print("Deep Learning Results")
print("=" * 70)

print("Shape :", dnn_results.shape)
print("\nColumns:")
print(list(dnn_results.columns))

display(dnn_results)

# ----------------------------------------------------------
# Missing Value Check
# ----------------------------------------------------------

print("\n")
print("=" * 70)
print("Missing Value Summary")
print("=" * 70)

ml_missing = ml_results.isnull().sum()
dnn_missing = dnn_results.isnull().sum()

missing_summary = pd.DataFrame({
    "ML Results": ml_missing,
    "DNN Results": dnn_missing
}).fillna("-")

display(missing_summary)

# ----------------------------------------------------------
# Data Types
# ----------------------------------------------------------

print("\n")
print("=" * 70)
print("Machine Learning Data Types")
print("=" * 70)

display(ml_results.dtypes)

print("\n")
print("=" * 70)
print("Deep Learning Data Types")
print("=" * 70)

display(dnn_results.dtypes)

# ----------------------------------------------------------
# Save Column Information
# ----------------------------------------------------------

column_info = pd.DataFrame({
    "ML Columns": pd.Series(ml_results.columns),
    "DNN Columns": pd.Series(dnn_results.columns)
})

column_info.to_csv(
    os.path.join(RESULT_DIR, "Result_File_Columns.csv"),
    index=False
)

print("\n✓ Results loaded successfully.")
print("✓ Column information saved.")
```

```python
ml_results = pd.read_csv(ML_RESULT_FILE)
dnn_results = pd.read_csv(DNN_RESULT_FILE)

print("ML Columns")
print(ml_results.columns.tolist())

print("\nDNN Columns")
print(dnn_results.columns.tolist())
```

```python
# ==========================================================
# Section 3 : ML vs DNN Comparison Table
# ==========================================================

# ----------------------------------------------------------
# Standardize DNN Column Names
# ----------------------------------------------------------

dnn_results = dnn_results.rename(columns={
    "F1-Score": "F1",
    "ROC-AUC": "ROC_AUC"
})

# ----------------------------------------------------------
# Standardize ML Column Names (Safety)
# ----------------------------------------------------------

ml_results = ml_results.rename(columns={
    "F1-Score": "F1",
    "ROC-AUC": "ROC_AUC"
})

# ----------------------------------------------------------
# Required Columns
# ----------------------------------------------------------

required_columns = [
    "Model",
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]

ml_results = ml_results[required_columns]
dnn_results = dnn_results[required_columns]

# ----------------------------------------------------------
# Merge Results
# ----------------------------------------------------------

final_results = pd.concat(
    [ml_results, dnn_results],
    ignore_index=True
)

# ----------------------------------------------------------
# Round Metrics
# ----------------------------------------------------------

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]

final_results[metric_columns] = final_results[
    metric_columns
].round(4)

# ----------------------------------------------------------
# Rank Models
# ----------------------------------------------------------

final_results = final_results.sort_values(
    by=["Accuracy", "ROC_AUC"],
    ascending=False
).reset_index(drop=True)

final_results.insert(
    0,
    "Rank",
    range(1, len(final_results)+1)
)

# ----------------------------------------------------------
# Display Final Table
# ----------------------------------------------------------

print("="*75)
print("FINAL MODEL COMPARISON")
print("="*75)

display(final_results)

# ----------------------------------------------------------
# Save Final Comparison
# ----------------------------------------------------------

final_file = os.path.join(
    RESULT_DIR,
    "Final_Model_Comparison.csv"
)

final_results.to_csv(
    final_file,
    index=False
)

print("\n✓ Final comparison table saved.")

print(final_file)
```

```python
# ==========================================================
# Section 4 : Accuracy Comparison
# ==========================================================

# ----------------------------------------------------------
# Sort Models by Accuracy
# ----------------------------------------------------------

accuracy_df = final_results.sort_values(
    by="Accuracy",
    ascending=False
)

# ----------------------------------------------------------
# Create Figure
# ----------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    accuracy_df["Model"],
    accuracy_df["Accuracy"],
    edgecolor="black",
    linewidth=0.8
)

# ----------------------------------------------------------
# Value Labels
# ----------------------------------------------------------

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.003,
        f"{height:.4f}",
        ha="center",
        fontsize=9
    )

# ----------------------------------------------------------
# Figure Formatting
# ----------------------------------------------------------

plt.title("Accuracy Comparison of Machine Learning and Deep Learning Models",
          fontsize=13,
          fontweight="bold")

plt.xlabel("Models")
plt.ylabel("Accuracy")

plt.xticks(rotation=30, ha="right")

plt.ylim(
    max(0.50, accuracy_df["Accuracy"].min() - 0.05),
    min(1.02, accuracy_df["Accuracy"].max() + 0.05)
)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# ----------------------------------------------------------
# Save Figure
# ----------------------------------------------------------

figure_path = os.path.join(
    FIGURE_DIR,
    "Accuracy_Comparison.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(f"✓ Figure saved to:\n{figure_path}")
```

```python
# ==========================================================
# Section 5 : Overall Performance Comparison
# ==========================================================

# Performance metrics
metrics = ["Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]

# Use model names as index
comparison_df = final_results.set_index("Model")[metrics]

# ----------------------------------------------------------
# Plot
# ----------------------------------------------------------

ax = comparison_df.plot(
    kind="bar",
    figsize=(12, 6),
    width=0.8
)

# ----------------------------------------------------------
# Formatting
# ----------------------------------------------------------

plt.title(
    "Performance Comparison of Machine Learning and Deep Learning Models",
    fontsize=13,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Score")

plt.xticks(rotation=30, ha="right")

plt.ylim(0.0, 1.05)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.legend(
    title="Metrics",
    loc="lower right",
    frameon=True
)

plt.tight_layout()

# ----------------------------------------------------------
# Save Figure
# ----------------------------------------------------------

figure_path = os.path.join(
    FIGURE_DIR,
    "Overall_Performance_Comparison.png"
)

plt.savefig(
    figure_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(f"✓ Figure saved to:\n{figure_path}")
```

```python
# ==========================================================
# Section 6 : Publication-Ready Final Results Table
# ==========================================================

# Copy results
publication_table = final_results.copy()

# ----------------------------------------------------------
# Round Performance Metrics
# ----------------------------------------------------------

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]

publication_table[metric_columns] = (
    publication_table[metric_columns]
    .astype(float)
    .round(4)
)

# ----------------------------------------------------------
# Display Table
# ----------------------------------------------------------

print("=" * 90)
print("Publication-Ready Model Comparison Table")
print("=" * 90)

display(publication_table)

# ----------------------------------------------------------
# Save CSV
# ----------------------------------------------------------

csv_path = os.path.join(
    RESULT_DIR,
    "Final_Model_Comparison.csv"
)

publication_table.to_csv(
    csv_path,
    index=False
)

# ----------------------------------------------------------
# Save Excel
# ----------------------------------------------------------

excel_path = os.path.join(
    RESULT_DIR,
    "Final_Model_Comparison.xlsx"
)

publication_table.to_excel(
    excel_path,
    index=False
)

# ----------------------------------------------------------
# Save LaTeX Table
# ----------------------------------------------------------

latex_path = os.path.join(
    RESULT_DIR,
    "Final_Model_Comparison_LaTeX.txt"
)

latex_table = publication_table.to_latex(
    index=False,
    float_format="%.4f",
    caption="Performance comparison of machine learning and deep learning models.",
    label="tab:model_comparison"
)

with open(latex_path, "w") as f:
    f.write(latex_table)

# ----------------------------------------------------------
# Summary
# ----------------------------------------------------------

print("\nFiles Saved Successfully")
print("-" * 40)
print(csv_path)
print(excel_path)
print(latex_path)
```

```python
# ==========================================================
# Section 7 : Best Model Analysis
# ==========================================================

best_model = final_results.iloc[0]

print("="*70)
print("BEST PERFORMING MODEL")
print("="*70)

print(f"Model      : {best_model['Model']}")
print(f"Accuracy   : {best_model['Accuracy']:.4f}")
print(f"Precision  : {best_model['Precision']:.4f}")
print(f"Recall     : {best_model['Recall']:.4f}")
print(f"F1-Score   : {best_model['F1']:.4f}")
print(f"ROC-AUC    : {best_model['ROC_AUC']:.4f}")

best_model_df = pd.DataFrame(best_model).T

display(best_model_df)

best_model_df.to_csv(
    os.path.join(RESULT_DIR, "Best_Model_Summary.csv"),
    index=False
)

print("\n✓ Best model summary saved.")
```

```python
# ==========================================================
# Section 8 : Final Performance Dashboard
# ==========================================================

metrics = ["Accuracy","Precision","Recall","F1","ROC_AUC"]

dashboard = final_results.set_index("Model")[metrics]

fig = plt.figure(figsize=(12,7))

plt.imshow(
    dashboard,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Performance")

plt.xticks(
    range(len(metrics)),
    metrics,
    rotation=0
)

plt.yticks(
    range(len(dashboard.index)),
    dashboard.index
)

plt.title(
    "Performance Dashboard of All Models",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()

dashboard_path = os.path.join(
    FIGURE_DIR,
    "Performance_Dashboard.png"
)

plt.savefig(
    dashboard_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("✓ Dashboard saved.")
```
